```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
df = pd.read_excel("../Upraveny_dataset.xlsx")
```

```python
stlpce = [
    # Demografia
    'Pohlavie', 'Vek_kat', 'vlna',
    
    # Komorbidity
    'Hypertenzia', 'Diabetes mellitus', 'KardiovaskulÃ¡rne ochorenia',
    'ChronickÃ© respiraÄnÃ© ochorenia', 'RenÃ¡lne ochorenia',
    'PeÄeÅˆovÃ© ochorenia', 'OnkologickÃ© ochorenia',
    
    # Lieky
    'MD652 | FABIFLU TABLETS', 'MD656 IV-BECT 6MG (ivermectin)', '5042D | VEKLURY', '9547D | PAXLOVID', 'LAGEVRIO', '00584 | PYRIDOXIN LÃ‰ÄŒIVA INJ', 'Vitamin C', 'Vitamin D', '00498 | MAGNESIUM SULFURICUM BBP 100 MG/ML INJEKÄŒNÃ ROZTOK', '00449 | EREVIT 300 MG/ML', 'Prednison', 'Dexametazon', '2410B HYDROCORTISONE', '3242C | OLUMIANT 4 MG', 'Kineret', 'RoActemra', '34045 | POLYOXIDONIUM 6 MG', '87299 | IMUNOR', '56930 IMMODIN', 'Isoprinosine/INOMED', '35715 Azithromycin', '45954 Ceftriaxon', 'Moxifloxacin', 'Ciprofloxacin', 'PPI', '94918 AMBROBENE', '24859 PENTOXYPHILLINUM', '8893 ACC INJEKT', '24949 CODEIN ', '26846 OXANTIL', 'Antikoagulancia', 'Antiagregacne',
    
    # ZÃ¡vislÃ¡ premennÃ¡
    'ZÃ¡vaÅ¾nosÅ¥ priebehu ochorenia'
]
```

```python
kategoricke = ['Pohlavie', 'Vek_kat', 'vlna', 'ZÃ¡vaÅ¾nosÅ¥ priebehu ochorenia']

# Pouzijeme iba analyzovane stlpce; 'Meno' sa tak nikdy nedostane do Apriori
df_bin = pd.get_dummies(df[stlpce], columns=kategoricke)

# Premenovanie dummy columns pre Vek_kat
vek_columns = {col: col.replace('Vek_kat_', 'VekovÃ¡ kategÃ³ria ') for col in df_bin.columns if col.startswith('Vek_kat_')}
df_bin = df_bin.rename(columns=vek_columns)

# Premenovanie dummy columns pre Pohlavie
pohlavie_columns = {col: col.replace('Pohlavie_', 'Pohlavie - ') for col in df_bin.columns if col.startswith('Pohlavie_')}
df_bin = df_bin.rename(columns=pohlavie_columns)

# Premenovanie dummy columns pre vlny
vlna_columns = {col: col.replace('vlna_', 'Vlna - ') for col in df_bin.columns if col.startswith('vlna_')}
df_bin = df_bin.rename(columns=vlna_columns)

# Premenovanie dummy columns pre ZÃ¡vaÅ¾nosÅ¥ priebehu ochorenia
zavaznost_columns = {col: col.replace('ZÃ¡vaÅ¾nosÅ¥ priebehu ochorenia_', 'VÃ½sledok hospitalizÃ¡cie - ') for col in df_bin.columns if col.startswith('ZÃ¡vaÅ¾nosÅ¥ priebehu ochorenia_')}
df_bin = df_bin.rename(columns=zavaznost_columns)
```

```python
binarne_stlpce = [
    'Hypertenzia', 'Diabetes mellitus', 'KardiovaskulÃ¡rne ochorenia',
    'ChronickÃ© respiraÄnÃ© ochorenia', 'RenÃ¡lne ochorenia',
    'PeÄeÅˆovÃ© ochorenia', 'OnkologickÃ© ochorenia',
    'MD652 | FABIFLU TABLETS', 'MD656 IV-BECT 6MG (ivermectin)', '5042D | VEKLURY', '9547D | PAXLOVID', 'LAGEVRIO', '00584 | PYRIDOXIN LÃ‰ÄŒIVA INJ', 'Vitamin C', 'Vitamin D', '00498 | MAGNESIUM SULFURICUM BBP 100 MG/ML INJEKÄŒNÃ ROZTOK', '00449 | EREVIT 300 MG/ML', 'Prednison', 'Dexametazon', '2410B HYDROCORTISONE', '3242C | OLUMIANT 4 MG', 'Kineret', 'RoActemra', '34045 | POLYOXIDONIUM 6 MG', '87299 | IMUNOR', '56930 IMMODIN', 'Isoprinosine/INOMED', '35715 Azithromycin', '45954 Ceftriaxon', 'Moxifloxacin', 'Ciprofloxacin', 'PPI', '94918 AMBROBENE', '24859 PENTOXYPHILLINUM', '8893 ACC INJEKT', '24949 CODEIN ', '26846 OXANTIL', 'Antikoagulancia', 'Antiagregacne'
]
```

```python
for stlpec in binarne_stlpce:
    if df_bin[stlpec].dtype == object:
        df_bin[stlpec] = df_bin[stlpec].map({'Ãno': 1, 'Nie': 0, 'Yes': 1, 'No': 0})

df_bin = df_bin.astype(bool)
```

```python
import time
start_time = time.time()

# -----------------------------------------------
# KROK 3: Apriori algoritmus
# -----------------------------------------------
itemsets = apriori(
    df_bin,
    min_support=0.10,      # kombinÃ¡cia sa musÃ­ vyskytovaÅ¥ aspoÅˆ v 10% prÃ­padov
    use_colnames=True,
    max_len=4              # max 4 poloÅ¾ky v pravidle (3 antecedenty + 1 consequent)
)

# -----------------------------------------------
# KROK 4: Generovanie pravidiel
# -----------------------------------------------
pravidla = association_rules(
    itemsets,
    metric="confidence",
    min_threshold=0.3      # minimÃ¡lna spoÄ¾ahlivosÅ¥ 30%
)

# Filtrovanie - iba pravidlÃ¡ s 1 consequent
pravidla = pravidla[pravidla['consequents'].apply(lambda x: len(x) == 1)]

# -----------------------------------------------
# KROK 5: Filtrovanie na jednotlivÃ© vÃ½sledky
# -----------------------------------------------
pravidla_vysledok_1 = pravidla[
    pravidla['consequents'].apply(
        lambda x: x == frozenset({'VÃ½sledok hospitalizÃ¡cie - 1'})
    )
].sort_values('lift', ascending=False)

pravidla_vysledok_2 = pravidla[
    pravidla['consequents'].apply(
        lambda x: x == frozenset({'VÃ½sledok hospitalizÃ¡cie - 2'})
    )
].sort_values('lift', ascending=False)

pravidla_vysledok_3 = pravidla[
    pravidla['consequents'].apply(
        lambda x: x == frozenset({'VÃ½sledok hospitalizÃ¡cie - 3'})
    )
].sort_values('lift', ascending=False)

# -----------------------------------------------
# KROK 6: Export vÃ½sledkov
# -----------------------------------------------
export = pd.concat([pravidla_vysledok_1, pravidla_vysledok_2, pravidla_vysledok_3])
export = export[export['lift'] >= 1.00]
export.to_excel('asociacne_pravidla.xlsx', index=False)

end_time = time.time()
print(f"Export hotovÃ½! Celkovo {len(export)} pravidiel uloÅ¾enÃ½ch do asociacne_pravidla.xlsx.")
print(f"VÃ½poÄet a export (Apriori) trval: {end_time - start_time:.2f} sekÃºnd.")
```

```python
import time
start_time_fp = time.time()

# -----------------------------------------------
# FP-GROWTH - alternatÃ­vny algoritmus
# -----------------------------------------------
from mlxtend.frequent_patterns import fpgrowth

itemsets_fp = fpgrowth(
    df_bin,
    min_support=0.10,
    use_colnames=True,
    max_len=4
)

pravidla_fp = association_rules(
    itemsets_fp,
    metric="confidence",
    min_threshold=0.3
)

# Filtrovanie na jednotlivÃ© vÃ½sledky
pravidla_fp_vysledok_1 = pravidla_fp[
    pravidla_fp['consequents'].apply(
        lambda x: x == frozenset({'VÃ½sledok hospitalizÃ¡cie - 1'})
    )
].sort_values('lift', ascending=False)

pravidla_fp_vysledok_2 = pravidla_fp[
    pravidla_fp['consequents'].apply(
        lambda x: x == frozenset({'VÃ½sledok hospitalizÃ¡cie - 2'})
    )
].sort_values('lift', ascending=False)

pravidla_fp_vysledok_3 = pravidla_fp[
    pravidla_fp['consequents'].apply(
        lambda x: x == frozenset({'VÃ½sledok hospitalizÃ¡cie - 3'})
    )
].sort_values('lift', ascending=False)

export_fp = pd.concat([pravidla_fp_vysledok_1, pravidla_fp_vysledok_2, pravidla_fp_vysledok_3])
export_fp = export_fp[export_fp['lift'] >= 1.00]
export_fp.to_excel('asociacne_pravidla_fpgrowth.xlsx', index=False)

end_time_fp = time.time()
print(f"Export hotovÃ½! Celkovo {len(export_fp)} pravidiel uloÅ¾enÃ½ch do asociacne_pravidla_fpgrowth.xlsx.")
print(f"VÃ½poÄet a export (FP-Growth) trval: {end_time_fp - start_time_fp:.2f} sekÃºnd.")
```
